## Ejercicio 16 - DataFrame Estudiantes

```python
import pandas as pd

datos = {
    'Nombre': ['Carlos', 'María', 'José', 'Ana', 'Luis', 'Sofía', 'Pedro', 'Valeria', 'Andrés', 'Camila'],
    'Carrera': ['Ingeniería', 'Medicina', 'Arquitectura', 'Derecho', 'Contaduría',
                'Psicología', 'Informática', 'Marketing', 'Biología', 'Diseño'],
    'Edad': [20, 22, 21, 23, 24, 19, 22, 20, 21, 23],
    'Promedio': [85, 92, 78, 88, 75, 95, 81, 79, 84, 90],
    'Ciudad': ['Tegucigalpa', 'San Pedro Sula', 'La Ceiba', 'Choluteca', 'Comayagua',
               'Danlí', 'El Progreso', 'Juticalpa', 'Santa Rosa', 'Puerto Cortés']
}

df = pd.DataFrame(datos)

print("Primeras 5 filas:")
print(df.head())

print("\nInformación general:")
df.info()

promedio_general = df['Promedio'].mean()
print(f"\nPromedio general de calificaciones: {promedio_general:.2f}")

print("\nEstudiantes con promedio mayor o igual a 80:")
print(df[df['Promedio'] >= 80])

print("\nEstudiantes ordenados por promedio de mayor a menor:")
print(df.sort_values(by='Promedio', ascending=False))
```

### Explicación
Este código usa Pandas para organizar información de estudiantes en una tabla. Primero guarda los datos de 10 estudiantes, luego muestra una vista rápida de la información, enseña detalles generales del DataFrame, calcula el promedio general de notas, filtra a los estudiantes con promedio de 80 o más, y al final los ordena de mayor a menor según su calificación.

## Ejercicio 17 - DataFrame Análisis de ventas con pandas

```python
import pandas as pd

datos = {
    'Fecha': ['2026-06-01', '2026-06-02', '2026-06-03', '2026-06-04', '2026-06-05',
              '2026-06-06', '2026-06-07', '2026-06-08', '2026-06-09', '2026-06-10',
              '2026-06-11', '2026-06-12', '2026-06-13', '2026-06-14', '2026-06-15'],
    
    'Vendedor': ['Carlos', 'María', 'José', 'Ana', 'Luis',
                 'Carlos', 'María', 'José', 'Ana', 'Luis',
                 'Carlos', 'María', 'José', 'Ana', 'Luis'],
    
    'Producto': ['Laptop', 'Mouse Gamer', 'Monitor', 'Teclado Mecánico', 'Auriculares',
                 'Tablet', 'Smartphone', 'Laptop', 'Mouse Gamer', 'Monitor',
                 'Smartwatch', 'Tablet', 'Teclado Mecánico', 'Auriculares', 'Smartphone'],
    
    'Categoría': ['Computadoras', 'Accesorios', 'Pantallas', 'Accesorios', 'Audio',
                  'Dispositivos', 'Celulares', 'Computadoras', 'Accesorios', 'Pantallas',
                  'Wearables', 'Dispositivos', 'Accesorios', 'Audio', 'Celulares'],
    
    'Cantidad': [2, 5, 3, 4, 6, 2, 3, 1, 7, 2, 4, 3, 2, 5, 2],
    
    'Precio Unitario': [850, 40, 250, 90, 60, 400, 700, 850, 40, 250, 300, 400, 90, 60, 700]
}

df = pd.DataFrame(datos)

df['TotalVenta'] = df['Cantidad'] * df['Precio Unitario']

print("DataFrame de ventas:")
print(df)

print("\nTotal vendido por vendedor:")
print(df.groupby('Vendedor')['TotalVenta'].sum())

print("\nProducto más vendido:")
print(df.groupby('Producto')['Cantidad'].sum().sort_values(ascending=False).head(1))

print("\nCategoría con mayor ingreso:")
print(df.groupby('Categoría')['TotalVenta'].sum().sort_values(ascending=False).head(1))

print("\nVentas ordenadas de mayor a menor:")
print(df.sort_values(by='TotalVenta', ascending=False))
```

### Explicación
Este código crea una tabla con ventas de productos tecnológicos, calcula el total de cada venta y luego analiza los datos para ver cuánto vendió cada vendedor, cuál producto se vendió más y qué categoría generó más dinero. Al final ordena las ventas de mayor a menor.

## Ejercicio 18 - Limpieza de datos

```python
import pandas as pd

datos = {
    'Nombre': ['Daniel', 'Fernanda', 'Kevin', 'Melissa', 'Jorge', 'Andrea', 'Samuel', 'Paola', 'Ricardo', 'Gabriela'],
    'Edad': [26, 29, None, 31, 24, 27, None, 30, 28, 23],
    'Correo': ['daniel@gmail.com', 'fernanda@gmail.com', None, 'melissa@gmail.com', 'jorge@gmail.com',
               None, 'samuel@gmail.com', 'paola@gmail.com', None, 'gabriela@gmail.com'],
    'Ciudad': ['Tegucigalpa', '', 'San Pedro Sula', None, 'La Ceiba',
               '', 'Comayagua', 'Choluteca', '', 'Danlí'],
    'Teléfono': ['9999-1234', None, '9888-2345', '9777-3456', None,
                 '9666-4567', '9555-5678', None, '9444-6789', '9333-7890']
}

df = pd.DataFrame(datos)

print("Valores nulos:")
print(df.isnull())

print("\nCantidad de valores nulos por columna:")
print(df.isnull().sum())

df['Ciudad'] = df['Ciudad'].replace('', 'No especificada')
df['Ciudad'] = df['Ciudad'].fillna('No especificada')

df = df.dropna(subset=['Correo'])

print("\nDataFrame limpio:")
print(df)
```

### Explicacion
Este código crea una tabla con datos incompletos de clientes, identifica los valores vacíos, los cuenta, reemplaza ciudades faltantes por “No especificada” y elimina registros sin correo para dejar el DataFrame limpio.

## Ejercicio 19 - Análisis de préstamos

```python
import pandas as pd

datos = {
    'Cliente': ['Carlos', 'María', 'José', 'Ana', 'Luis', 'Sofía', 'Pedro', 'Valeria', 'Andrés', 'Camila'],
    'Monto Solicitado': [45000, 60000, 30000, 75000, 52000, 40000, 85000, 25000, 68000, 55000],
    'Plazo (meses)': [12, 24, 6, 36, 18, 12, 48, 6, 24, 18],
    'Ingreso Mensual': [20000, 35000, 15000, 40000, 28000, 22000, 50000, 18000, 32000, 30000],
    'Estado': ['Aprobado', 'Pendiente', 'Rechazado', 'Aprobado', 'Pendiente',
               'Aprobado', 'Aprobado', 'Rechazado', 'Pendiente', 'Aprobado']
}

df = pd.DataFrame(datos)

promedio_monto = df['Monto Solicitado'].mean()
print("Monto promedio solicitado:")
print(promedio_monto)

print("\nSolicitudes por estado:")
print(df['Estado'].value_counts())

print("\nPréstamos aprobados:")
print(df[df['Estado'] == 'Aprobado'])

print("\nClientes con monto solicitado mayor a 50,000:")
print(df[df['Monto Solicitado'] > 50000])
```

### Explicación
Este código crea una tabla de solicitudes de préstamo y analiza los datos calculando el promedio solicitado, contando solicitudes por estado, mostrando préstamos aprobados, clientes con montos mayores a 50,000 y el promedio por cada estado.

## Ejercicio 20 - Mini análisis de datos tipo empresa

```python
import pandas as pd

datos = {
    'Nombre': ['Daniel', 'Fernanda', 'Kevin', 'Melissa', 'Jorge', 'Andrea', 'Samuel', 'Paola', 'Ricardo', 'Gabriela'],
    'Departamento': ['TI', 'Recursos Humanos', 'Ventas', 'Finanzas', 'TI', 'Marketing', 'Ventas', 'Finanzas', 'TI', 'Marketing'],
    'Cargo': ['Analista', 'Coordinadora', 'Ejecutivo', 'Contadora', 'Desarrollador', 'Diseñadora', 'Supervisor', 'Analista', 'Ingeniero', 'Gerente'],
    'Salario': [25000, 30000, 22000, 28000, 35000, 24000, 32000, 27000, 40000, 45000],
    'Años de Experiencia': [2, 6, 4, 8, 5, 3, 7, 6, 10, 12],
    'Ciudad': ['Tegucigalpa', 'San Pedro Sula', 'La Ceiba', 'Comayagua', 'Tegucigalpa',
               'Choluteca', 'El Progreso', 'Danlí', 'Tegucigalpa', 'San Pedro Sula']
}

df = pd.DataFrame(datos)

df['NivelExperiencia'] = df['Años de Experiencia'].apply(
    lambda x: 'Junior' if x <= 3 else 'Intermedio' if x <= 7 else 'Senior'
)

print("Primeros registros:")
print(df.head())

print("\nEstadísticas generales de salarios:")
print("Salario promedio:", df['Salario'].mean())
print("Salario mínimo:", df['Salario'].min())
print("Salario máximo:", df['Salario'].max())

print("\nSalario promedio por departamento:")
print(df.groupby('Departamento')['Salario'].mean())

print("\nEmpleado con salario más alto:")
print(df[df['Salario'] == df['Salario'].max()])

print("\nEmpleados con más de 5 años de experiencia:")
print(df[df['Años de Experiencia'] > 5])

print("\nEmpleados ordenados por salario:")
print(df.sort_values(by='Salario', ascending=False))
```

## Explicación
Este código analiza datos de empleados, mostrando salarios, experiencia, promedios por departamento y clasificando a cada empleado según su nivel de experiencia.

## 1. ¿Que datos genera?
Spotify genera millones de datos relacionados con el comportamiento de sus usuarios y el contenido musical.

### Tipos de datos:

- Canciones Reproducidas
- Tiempo de escucha
- Generos musicales preferidos
- Historial de busqueda
- Artistas Favoritas
- Ubicacion geografica del usuario
- Hora y frecuencia de uso
- Interacciones de anuncios (en cuentas con plan gratuito)

### Volumen:
Spotify tiene cientos de millones de usuarios en el mundo, generando millones de reproducciones al dia, lo que representa petabytes de datos.

### Frecuencia:
Los datos se generan en tiempo real cada vez que un usuario:
- reproduce musica,
- cambia cancion,
- pausa contenido,
- crea playlists
- interactua con recomendaciones.

## 2. ¿Quien los usa?
Los datos son utilizados por diferentes actores:

### Empresa (Spotify)
- recomendar musica personalizada,
- crear playlist automaticas,
- mejorar la experiencia del usuario,
- analizar tendencias musicales.

### Artistas y productores
- conocer que canciones son mas escuchadas,
- identificar mercados con mayor audiencia,
- planificar lanzamientos o giras.

### Investigadores y analistas
- estudiar patrones de consumo digital,
- comportamiento del usuario,
- tendencias culturales y musicales.

## 3. ¿Que problema resuelve?
Spotify utiliza Big Data para resolver el problema de encontrar contenido relevante dentro de un catalogo extremadamente grande.

Sin el analisis de datos, seria imposible para un usuario descrubrir musica nueva entre miles de canciones.

Al implementar el uso de Big Data permite:
- personalizar recomendaciones musicales,
- mejorar la experiencia del usuario,
- ayudar a artistas a comprender su audiencia,
- optimizar campañas publicitarias.

## 4. ¿Por que no bastaria Excel?
Sabemos que Excel llega a tener un limite para la entrada de datos, llegara un momento donde colapsara y ya no dejara ingresar mas. Ya que, Spotify procesa millones de interacciones diariamente. 

Los datos se generan constantemente en tiempo real para que Spotify las procese de inmediato y asi actualizar recomendaciones y estadisticas.

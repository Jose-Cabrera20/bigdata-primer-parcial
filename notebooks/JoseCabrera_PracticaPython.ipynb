## Ejercicio 7 - Promedio de calificaciones

```python
def calcular_promedio(calificaciones):
    suma = sum(calificaciones)
    prom = suma / len(calificaciones)

    if prom >= 70:
        resultado = "Aprobó"
    else:
        resultado = "Reprobó"

    return prom, resultado


calificaciones = [100, 72, 80, 40, 100]

prom, resultado = calcular_promedio(calificaciones)

print(f"Calificaciones: {calificaciones}")
print(f"Promedio: {prom}")
print(f"Resultado: {resultado}")
```

### Explicacion
Se hizo una suma de calificaciones para poder sacar el promedio con una evaluacion con if si el promedio es mayor a 70.

## Ejercicio 8 - Lista de numeros positivos y negativos

```python
def cnumeros(lista):
    ceros = 0
    positivos = 0
    negativos = 0

    for numero in lista:
        if numero > 0:
            positivos += 1
        elif numero < 0:
            negativos += 1
        else:
            ceros += 1

    return ceros, positivos, negativos


numero = [10, -4, 80, 93, 0, 0, -65, 93]

ceros, positivos, negativos = cnumeros(numero)

print(f"Números: {numero}")
print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Ceros: {ceros}")
```

### Explicacion
Se genera la lista de numeros y luego los cuenta y define cuales son los positivos, negativos y ceros por medio de if, elif y else.

## Ejercicio 9 - Mayor y menor de una lista

```python
def analisis_lista(lista):
    mayor = max (lista)
    menor = min (lista)
    diferencia = mayor - menor
    return mayor, menor, diferencia

numeros = [1,2,3,8,19,4,7,30,80,41,40,37,10]

mayor, menor, diferencia = analisis_lista(numeros)

print(f"Lista: {numeros}")
print(f"Numero Mayor: {mayor}")
print(f"Numero Menor: {menor}")
print(f"Diferencia: {diferencia}")
```

### Explicacion
Se utilizaron las funciones de max() y min() para poder sacar de una lista el numero mayor y menor y sacar la diferencia entre ellos.

## Ejemplo 12 - Validacion de contraseña

```python
password = "holamundo1329"

valida = (
    len(password) >= 8 and
    any(c.isupper() for c in password) and
    any(c.islower() for c in password) and
    any(c.isdigit() for c in password)
)

print("Contraseña valida" if valida else "Contraseña invalida")
```

## Explicacion 
se evalua la contraseña con las especificaciones de contraseña que pide.

## Ejercicio 13 - Sistema de Inventario

```python
productos = [
    {"codigo": 1, "nombre": "Frijoles", "precio": 50, "cantidad": 7},
    {"codigo": 2, "nombre": "Pan Molde", "precio": 70, "cantidad": 9}
]

for p in productos:
    print(p)

total = sum(p["precio"] * p["cantidad"] for p in productos)
print("Valor total:", total)
```

## Ejercicio 14 - Registro de ventas


```python
ventas = [
    {"cliente": "Jose", "producto": "Frijoles", "cantidad": 2, "precio_unitario": 50},
    {"cliente": "Karina", "producto": "Pan Molde", "cantidad": 7, "precio_unitario": 70}
]

total_general = 0
compra_mas_alta = 0
cliente_mayor_compra = ""

print("Registro de ventas")

for venta in ventas:
    total_venta = venta["cantidad"] * venta["precio_unitario"]
    total_general += total_venta

    print(f"Cliente: {venta['cliente']}")
    print(f"Producto: {venta['producto']}")
    print(f"Total venta: L. {total_venta}")
    print()

    if total_venta > compra_mas_alta:
        compra_mas_alta = total_venta
        cliente_mayor_compra = venta["cliente"]

print(f"Total general vendido: L. {total_general}")
print(f"Cliente con la compra más alta: {cliente_mayor_compra}")
print(f"Monto de la compra más alta: L. {compra_mas_alta}")
```

## Explicacion
Se usa una lista de diccionarios para guardar los datos de cada venta. Luego, mediante un ciclo, se calcula el total correspondiente a cada venta y se va sumando al total acumulado general. Además, se evalúa cada registro para identificar cuál venta tiene el monto más alto.

# Ejercicio 15 - Limpieza básica de texto

```python
frase = "      No hay nada mejor que la programacion con python   "

texto_limpio = frase.strip()
texto_limpio = texto_limpio.lower()
texto_limpio = texto_limpio.replace("  ", " ")

palabras = texto_limpio.split()

print(f"Texto original: '{frase}'")
print(f"Texto limpio: '{texto_limpio}'")
print(f"Cantidad de palabras: {len(palabras)}")
```

## Explicacion
se utilizan métodos de cadenas para limpiar y transformar el texto. strip() elimina espacios innecesarios al inicio y final, lower() convierte todo a minúsculas, replace() sustituye espacios dobles por uno simple y split() divide la frase en palabras para poder contarlas.

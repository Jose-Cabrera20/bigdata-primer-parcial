## Ejercicio 1 -  Clasificación de edad

```python
edad = 21

if edad < 13:
    categoria = "Niño"
elif edad <= 17:
    categoria = "Adolescente"
elif edad <= 59:
    categoria = "Adulto"
else:
    categoria = "Adulto mayor"

print(categoria)
```

### Explicacion

Se usan condiciones para clasificar la edad.

## Ejercicio 5 - Tabla de multiplicar

```python
numero = 2

print(f"Tabla de multiplicar del {numero}")

for i in range(1, 13):
    resultado = numero * i
    print(f"{numero} x {i} = {resultado}")
```

## Ejercicio 6 - Contador de vocales

```python
frase = "La clase de Big Data es grandiosa"

conta = 0
vocales = "aeiouAEIOU"

for letra in frase:
    if letra in vocales:
        conta += 1

print(f"Frase: {frase}")
print(f"Cantidad de vocales: {conta}")
```

### Explicacion
Se hizo un conteo de vocales de una frase con for y dentro de el teniendo un contador que suma las vocales.

## Ejercicio 7 - Promedio de calificaciones

```python
def calcular_promedio(calificaciones):
    suma = sum(calificaciones)
    prom = suma / len(calificaciones)

    if prom >= 70:
        resultado = "Aprobó"
    else:
        resultado = "Reprobó"

    return prom, resultado


calificaciones = [100, 72, 80, 40, 100]

prom, resultado = calcular_promedio(calificaciones)

print(f"Calificaciones: {calificaciones}")
print(f"Promedio: {prom}")
print(f"Resultado: {resultado}")
```

### Explicacion
Se hizo una suma de calificaciones para poder sacar el promedio con una evaluacion con if si el promedio es mayor a 70.
